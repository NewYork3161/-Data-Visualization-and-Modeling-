```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm

# Load dataset (adjust path if needed)
df = pd.read_csv(
    r"C:\Users\hudso\OneDrive\Desktop\Data_Visualization_And_Modeling_Online-main\Data_Visualization_And_Modeling_Online-main\Lab\Employee_productivity.csv"
)

df.head()
```

```python
# Visualization
sns.regplot(x='YearsAtCompany', y='Productivity', data=df)

plt.title('Productivity vs Years at Company')
plt.xlabel('Years at Company')
plt.ylabel('Productivity')
plt.show()

# Linear regression
X = sm.add_constant(df['YearsAtCompany'])
y = df['Productivity']

model_years = sm.OLS(y, X).fit()

print("R-squared (Years at Company):")
print(model_years.rsquared)
```

```python

# Create dummy variables for DeskLocation
df_desk = pd.get_dummies(df, columns=['DeskLocation'], drop_first=True)

# Define predictors (include quantitative + new desk variables)
X = df_desk[['HoursWorked',
             'FocusScore',
             'YearsAtCompany',
             'TeamMeetingsPerWeek',
             'DeskLocation_Private Office',
             'DeskLocation_Remote']]

X = sm.add_constant(X)

# Define target variable
y = df_desk['ProductivityScore']

# Fit model
model_desk = sm.OLS(y, X).fit()

print(model_desk.summary())
```

```python
df_desk = pd.get_dummies(df, columns=['DeskLocation'], drop_first=True)

# Define predictors
X = df_desk[['HoursWorked',
             'FocusScore',
             'YearsAtCompany',
             'TeamMeetingsPerWeek',
             'DeskLocation_Private Office',
             'DeskLocation_Remote']]

# Convert everything to numeric
X = X.astype(float)

# Add constant
X = sm.add_constant(X)

# Define target
y = df_desk['ProductivityScore'].astype(float)

# Fit model
model_desk = sm.OLS(y, X).fit()

print(model_desk.summary())
```

```python
# Create dummy variables for Department
df_dept = pd.get_dummies(df, columns=['Department'], drop_first=True)

# Define predictors (quantitative + department dummies)
X = df_dept.drop(columns=['ProductivityScore'])

# Keep only numeric columns
X = X.select_dtypes(include=['int64', 'float64'])

# Convert to float to avoid object errors
X = X.astype(float)

# Add constant
X = sm.add_constant(X)

# Define target variable
y = df_dept['ProductivityScore'].astype(float)

# Fit model
model_dept = sm.OLS(y, X).fit()

print(model_dept.summary())
```

After building several regression models, I examined how different variables affected ProductivityScore. What I discovered is that FocusScore consistently showed the strongest relationship with productivity. When I added HoursWorked, YearsAtCompany, TeamMeetingsPerWeek, DeskLocation, and Department into the model, the overall R-squared value did not increase by a meaningful amount beyond what FocusScore already explained.

As I reviewed the regression output, I noticed that FocusScore was statistically significant, while most of the other predictors were not. This led me to conclude that productivity in this dataset is driven much more by an employee’s ability to focus than by how long they have been at the company, how many hours they work, or where they sit.

Through this analysis, I found that adding categorical variables such as DeskLocation and Department did not substantially improve the model’s explanatory power. Based on what I observed, I would recommend that the company focus on strategies that improve employee concentration rather than emphasizing tenure, meeting frequency, or office layout.

Overall, I determined that FocusScore provides the clearest and most meaningful explanation of productivity in this dataset.

Once I finished working through the entire lab and reviewing each regression model, I stepped back and looked at the overall patterns in the data. I started by analyzing the simple linear relationships between YearsAtCompany, HoursWorked, and FocusScore against ProductivityScore. After calculating the R-squared values for each one, it became clear to me that FocusScore stood out immediately. The R-squared for YearsAtCompany and HoursWorked was extremely small, which showed me that those variables explained almost none of the variation in productivity. FocusScore, on the other hand, accounted for a substantial portion of the changes in ProductivityScore, indicating a strong linear relationship.

As I moved forward and expanded the analysis into a multiple regression model, I incorporated additional variables such as TeamMeetingsPerWeek and DeskLocation along with the original predictors. After examining the regression summary, I saw that FocusScore continued to remain statistically significant, while most of the other variables did not demonstrate meaningful statistical impact. Even though the overall R-squared increased slightly in the full model, I recognized that the improvement was minimal compared to what FocusScore alone had already explained.

When looking more closely at the regression coefficients and p-values, I realized that neither HoursWorked nor YearsAtCompany had a statistically significant effect on productivity. That led me to conclude that simply working more hours or having more experience at the company does not necessarily translate into higher performance. I also noticed that the number of meetings per week did not appear to significantly influence productivity in the model, despite employee concerns about excessive meetings. In one version of the model, remote desk location showed a negative association with productivity, which suggests to me that location may have some influence, but not nearly as strong as focus.

Taking everything together, I determined that focus is the dominant factor driving productivity in this dataset. Based on what I observed throughout the analysis, I would recommend that management concentrate on strategies that enhance employees’ ability to focus. Rather than encouraging longer hours or relying on tenure as an indicator of performance, I believe the company would benefit more from reducing distractions, limiting unnecessary meetings, and structuring the work environment to support sustained concentration.
